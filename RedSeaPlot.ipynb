```python
import matplotlib.pyplot as plt
import numpy as np

# 資料
x = np.array([2, 3, 4, 5])  # X軸
ours_y = [713580, 1517508, 2660952, 3257784]
multiHist_y = [930579, 10521452, 55861220, 94751808]
copulaBase_y = [1267200, 2109600, 3247200, 3988800]

# 畫線圖
plt.figure(figsize=(8, 5))
plt.plot(x, ours_y, marker='o', label='Ours')
plt.plot(x, multiHist_y, marker='s', label='multi-histogram')
plt.plot(x, copulaBase_y, marker='^', label='Copula Base')

# 設定標籤
plt.xlabel('Number of Variables')
plt.ylabel('Storage')
plt.title('Comparison of Storage')

# x軸顯示固定的刻度
plt.xticks(x)
#plt.yscale('log')
# 加上圖例與網格
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
#RedSea Self Study Different Bins

import numpy as np
import matplotlib.pyplot as plt

# Given data
bin64 = np.array([0.021, 0.022, 0.024, 0.026])
bin128 = np.array([0.024, 0.023, 0.025, 0.028])
bin256 = np.array([0.024, 0.024, 0.027, 0.030])

# X axis labels
variables = np.array([2, 3, 4, 5])

# Plot settings
width = 0.25  # bar width spacing

plt.figure(figsize=(10, 5), dpi=300)

# Offsets
plt.bar(variables - width, bin64, width, label="bins 64", color="#1f77b4")
plt.bar(variables, bin128, width, label="bins 128", color="#ff7f0e")
plt.bar(variables + width, bin256, width, label="bins 256", color="#2ca02c")

# Labels, ticks, legend
plt.xlabel("Number of Variables", fontsize=12)
plt.ylabel("EMD Value", fontsize=12)
plt.title("Red Sea: EMD Comparison under Different Bins Numbers", fontsize=14)
plt.xticks(variables)

# Optional grid
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Legend
plt.legend()

plt.tight_layout()
plt.show()
```
